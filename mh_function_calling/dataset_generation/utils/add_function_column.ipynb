```python
import json
import os
from pathlib import Path

# Directory containing JSON files
data_dir = "/home/jovyan/Works/MSc_Thesis/mh_function_calling/dataset_generation/fc_training_data"
```

```python
def add_function_name_to_file(file_path):
    """Add function name to each sample in a JSON file without changing encoding"""
    
    # Extract function name from filename
    filename = file_path.name
    base_name = filename.replace('_data.json', '').replace('.json', '')
    function_name = f"get_{base_name}_data"
    
    print(f"Processing {filename} -> {function_name}")
    
    # Load JSON data with explicit UTF-8 encoding
    with open(file_path, 'r', encoding='utf-8') as f:
        data = json.load(f)
    
    # Add function name to each sample
    for sample in data:
        sample['function'] = function_name
    
    # Save back to file with same encoding and preserve Unicode
    with open(file_path, 'w', encoding='utf-8') as f:
        json.dump(data, f, indent=4, ensure_ascii=False)
    
    print(f"✅ Added function name to {len(data)} samples")
```

```python
# Test with one file
test_file = Path(data_dir) / "body_fat_data.json"
add_function_name_to_file(test_file)
```

```python
# Process all JSON files in directory
json_files = list(Path(data_dir).glob("*.json"))

print(f"Found {len(json_files)} JSON files")

for json_file in json_files:
    add_function_name_to_file(json_file)

print("\n✅ All files processed!")
```
