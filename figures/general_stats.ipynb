```python
# 
disease = ["Anxiety","Depression","Bipolar","Disruptive Behaviour","Schizophrenia","Eating Disorders"]
num_people= [301,280,40,40, 24,14]
```

```python
import matplotlib.pyplot as plt

# Create a bar graph
plt.figure(figsize=(8, 6))
bars = plt.bar(disease, num_people, color=['#FF6B6B', '#ffe66d', '#FFEAA7', '#96CEB4', '#45B7D1', '#DDA0DD', '#FFB347'])

# Customize the plot
plt.title('People Affected by Mental Health Disorders Globally', fontsize=16, fontweight='bold')
plt.xlabel('Mental Health Disorder', fontsize=12)
plt.ylabel('Number of People (in million)', fontsize=12)

# Rotate x-axis labels for better readability
plt.xticks(rotation=45, ha='right',fontsize=11)

# Add value labels on top of each bar
for bar, value in zip(bars, num_people):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 2, 
             str(value), ha='center', va='bottom', fontweight='bold')

# Add grid for better readability
plt.grid(axis='y', alpha=0.3, linestyle='--')

# Adjust layout to prevent label cutoff
plt.tight_layout()

# Display the plot
plt.show()
```
